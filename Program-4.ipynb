```python
from nltk.corpus import wordnet as wn

# Define the words to be analyzed
words = ['bear', 'set', 'square', 'lead', 'criteria']

# Function to find the 3rd meaning of a word
def third_meaning(word):
    synsets = wn.synsets(word)
    if len(synsets) >= 3:
        return synsets[2]
    else:
        return None

# Function to extract nouns from the synonyms list
def extract_nouns(word):
    return wn.synsets(word, pos=wn.NOUN)

# Function to extract verbs from the synonyms list
def extract_verbs(word):
    return wn.synsets(word, pos=wn.VERB)

# Function to extract adjectives from the synonyms list
def extract_adjectives(word):
    return wn.synsets(word, pos=wn.ADJ)

# Function to extract adverbs from the synonyms list
def extract_adverbs(word):
    return wn.synsets(word, pos=wn.ADV)

# Function to extract the definition of a word
def word_definition(word):
    synsets = wn.synsets(word)
    if synsets:
        return synsets[0].definition()
    else:
        return None

# Function to find hypernyms of a word
def find_hypernyms(word):
    synsets = wn.synsets(word)
    hypernyms = []
    for synset in synsets:
        hypernyms.extend(synset.hypernyms())
    return hypernyms

# Function to find hyponyms of a word
def find_hyponyms(word):
    synsets = wn.synsets(word)
    hyponyms = []
    for synset in synsets:
        hyponyms.extend(synset.hyponyms())
    return hyponyms

# Function to find similarities between any two hyponyms of a word
def hyponym_similarities(word):
    hyponyms = find_hyponyms(word)
    if len(hyponyms) >= 2:
        return hyponyms[0].path_similarity(hyponyms[1])
    else:
        return None

for word in words:
    print(f"--- {word.upper()} ---")
    print(f"3rd Meaning: {third_meaning(word)}")
    print(f"Nouns: {extract_nouns(word)}")
    print(f"Verbs: {extract_verbs(word)}")
    print(f"Adjectives: {extract_adjectives(word)}")
    print(f"Adverbs: {extract_adverbs(word)}")
    print(f"Definition: {word_definition(word)}")
    print(f"Hypernyms: {find_hypernyms(word)}")
    print(f"Hyponyms: {find_hyponyms(word)}")
    print(f"Similarity between two hyponyms: {hyponym_similarities(word)}\n")

```
